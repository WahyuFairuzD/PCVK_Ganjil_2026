```python
import cv2
import matplotlib.pyplot as plt

img_ktp = cv2.imread(
    '/content/drive/MyDrive/PCVK/Images/ktp1.jpg',
    cv2.IMREAD_GRAYSCALE
)

if img_ktp is None:
    print("Gambar ktp1.jpg tidak ditemukan. Periksa lokasi file.")
else:
    print("Gambar berhasil dibaca!")
    print("Ukuran citra:", img_ktp.shape)

    plt.figure(figsize=(10, 5))
    plt.imshow(img_ktp, cmap='gray')
    plt.title('Citra KTP Awal')
    plt.axis('off')
    plt.show()
```

```python
# -------------------------------------------------------------------------
# 9. Top-hat Transformation
# -------------------------------------------------------------------------
img_tophat = cv2.imread(img_ktp_path, cv2.IMREAD_GRAYSCALE)

if img_tophat is not None:
    kernel = np.ones((5, 5), np.uint8)

    tophat = cv2.morphologyEx(
        img_tophat,
        cv2.MORPH_TOPHAT,
        kernel
    )

    plt.figure(figsize=(12, 4))
    plt.subplot(131)
    plt.imshow(img_tophat, cmap='gray')
    plt.title('Citra Awal')
    plt.axis('off')

    plt.subplot(132)
    plt.imshow(cv2.morphologyEx(
        img_tophat, cv2.MORPH_OPEN, kernel
    ), cmap='gray')
    plt.title('Hasil Opening')
    plt.axis('off')

    plt.subplot(133)
    plt.imshow(tophat, cmap='gray')
    plt.title('Hasil Top-hat')
    plt.axis('off')

    plt.tight_layout()
    plt.show()
else:
    print("Gambar ktp1.jpg tidak ditemukan.")
```

```python
# -------------------------------------------------------------------------
# 10. Black-hat Transformation
# -------------------------------------------------------------------------
img_blackhat = cv2.imread(img_ktp_path, cv2.IMREAD_GRAYSCALE)

if img_blackhat is not None:
    kernel = np.ones((5, 5), np.uint8)

    blackhat = cv2.morphologyEx(
        img_blackhat,
        cv2.MORPH_BLACKHAT,
        kernel
    )

    plt.figure(figsize=(12, 4))
    plt.subplot(131)
    plt.imshow(img_blackhat, cmap='gray')
    plt.title('Citra Awal')
    plt.axis('off')

    plt.subplot(132)
    plt.imshow(cv2.morphologyEx(
        img_blackhat, cv2.MORPH_CLOSE, kernel
    ), cmap='gray')
    plt.title('Hasil Closing')
    plt.axis('off')

    plt.subplot(133)
    plt.imshow(blackhat, cmap='gray')
    plt.title('Hasil Black-hat')
    plt.axis('off')

    plt.tight_layout()
    plt.show()
else:
    print("Gambar ktp1.jpg tidak ditemukan.")
```

```python
# -------------------------------------------------------------------------
# 11. Skeletonization
# -------------------------------------------------------------------------
from skimage.morphology import skeletonize

img_skeleton = cv2.imread(img_ktp_path, cv2.IMREAD_GRAYSCALE)

if img_skeleton is not None:
    _, binary = cv2.threshold(
        img_skeleton,
        0,
        255,
        cv2.THRESH_BINARY + cv2.THRESH_OTSU
    )

    # Skeletonize menggunakan citra biner
    binary_bool = binary > 0
    skeleton = skeletonize(binary_bool)

    plt.figure(figsize=(12, 4))

    plt.subplot(131)
    plt.imshow(img_skeleton, cmap='gray')
    plt.title('Citra Awal')
    plt.axis('off')

    plt.subplot(132)
    plt.imshow(binary, cmap='gray')
    plt.title('Citra Biner')
    plt.axis('off')

    plt.subplot(133)
    plt.imshow(skeleton, cmap='gray')
    plt.title('Hasil Skeletonization')
    plt.axis('off')

    plt.tight_layout()
    plt.show()
else:
    print("Gambar ktp1.jpg tidak ditemukan.")
```

```python
# -------------------------------------------------------------------------
# 12. Perbandingan Operasi Morfologi
# -------------------------------------------------------------------------
img_compare = cv2.imread(img_ktp_path, cv2.IMREAD_GRAYSCALE)

if img_compare is not None:
    kernel = np.ones((3, 3), np.uint8)

    erosion = cv2.erode(img_compare, kernel, iterations=1)
    dilation = cv2.dilate(img_compare, kernel, iterations=1)
    opening = cv2.morphologyEx(
        img_compare, cv2.MORPH_OPEN, kernel
    )
    closing = cv2.morphologyEx(
        img_compare, cv2.MORPH_CLOSE, kernel
    )

    plt.figure(figsize=(12, 8))

    images = [
        (img_compare, 'Citra Awal'),
        (erosion, 'Erosi'),
        (dilation, 'Dilasi'),
        (opening, 'Opening'),
        (closing, 'Closing')
    ]

    for i, (image, title) in enumerate(images, start=1):
        plt.subplot(2, 3, i)
        plt.imshow(image, cmap='gray')
        plt.title(title)
        plt.axis('off')

    plt.tight_layout()
    plt.show()
else:
    print("Gambar ktp1.jpg tidak ditemukan.")
```

```python
# -------------------------------------------------------------------------
# 13. Penerapan Morfologi pada Citra KTP
# -------------------------------------------------------------------------
img_ktp = cv2.imread(img_ktp_path, cv2.IMREAD_GRAYSCALE)

if img_ktp is not None:
    # 1. Mengubah citra menjadi biner menggunakan Otsu
    _, binary_ktp = cv2.threshold(
        img_ktp,
        0,
        255,
        cv2.THRESH_BINARY + cv2.THRESH_OTSU
    )

    # 2. Membuat kernel morfologi
    kernel = np.ones((3, 3), np.uint8)

    # 3. Mengurangi noise kecil menggunakan Opening
    opening_ktp = cv2.morphologyEx(
        binary_ktp,
        cv2.MORPH_OPEN,
        kernel,
        iterations=1
    )

    # 4. Menutup celah kecil menggunakan Closing
    closing_ktp = cv2.morphologyEx(
        binary_ktp,
        cv2.MORPH_CLOSE,
        kernel,
        iterations=1
    )

    # 5. Menampilkan hasil
    plt.figure(figsize=(14, 8))

    plt.subplot(2, 2, 1)
    plt.imshow(img_ktp, cmap='gray')
    plt.title('Citra KTP Asli')
    plt.axis('off')

    plt.subplot(2, 2, 2)
    plt.imshow(binary_ktp, cmap='gray')
    plt.title('Hasil Threshold Otsu')
    plt.axis('off')

    plt.subplot(2, 2, 3)
    plt.imshow(opening_ktp, cmap='gray')
    plt.title('Hasil Opening')
    plt.axis('off')

    plt.subplot(2, 2, 4)
    plt.imshow(closing_ktp, cmap='gray')
    plt.title('Hasil Closing')
    plt.axis('off')

    plt.tight_layout()
    plt.show()

else:
    print("Gambar ktp1.jpg tidak ditemukan. Periksa kembali path file.")
```

```python
# -------------------------------------------------------------------------
# 14. Menyimpan Hasil Morfologi
# -------------------------------------------------------------------------
output_path = '/content/drive/MyDrive/PCVK/Images/ktp1_morfologi.png'

if img_ktp is not None:
    cv2.imwrite(output_path, closing_ktp)
    print("Hasil morfologi disimpan di:")
    print(output_path)
```

Pada praktikum Modul 9, dilakukan pengolahan citra menggunakan operasi morfologi, yaitu dilasi, erosi, opening, dan closing. Selain itu, dilakukan pula transformasi Top-hat, Black-hat, serta skeletonization untuk mengamati perubahan bentuk dan struktur objek pada citra.

Pada citra KTP, thresholding digunakan untuk memisahkan objek dari latar belakang, sedangkan operasi opening dan closing digunakan untuk membantu mengurangi noise dan menutup celah kecil. Hasil setiap operasi dibandingkan dengan citra awal untuk mengamati perubahan yang terjadi.

Pemilihan operasi morfologi perlu disesuaikan dengan kondisi citra karena setiap operasi dapat menghasilkan perubahan yang berbeda. Hasil yang paling sesuai ditentukan berdasarkan pengamatan terhadap keterbacaan tulisan pada citra KTP.

```python

# -------------------------------------------------------------------------
# 17. Evaluasi Hasil Morfologi
# -------------------------------------------------------------------------
img_ktp = cv2.imread(img_ktp_path, cv2.IMREAD_GRAYSCALE)

if img_ktp is not None:
    _, binary_ktp = cv2.threshold(
        img_ktp,
        0,
        255,
        cv2.THRESH_BINARY + cv2.THRESH_OTSU
    )

    kernel = np.ones((3, 3), np.uint8)

    hasil_erosi = cv2.erode(binary_ktp, kernel, iterations=1)
    hasil_dilasi = cv2.dilate(binary_ktp, kernel, iterations=1)

    hasil_opening = cv2.morphologyEx(
        binary_ktp, cv2.MORPH_OPEN, kernel
    )

    hasil_closing = cv2.morphologyEx(
        binary_ktp, cv2.MORPH_CLOSE, kernel
    )

    daftar_hasil = [
        (img_ktp, 'Citra Awal'),
        (binary_ktp, 'Threshold Otsu'),
        (hasil_erosi, 'Erosi'),
        (hasil_dilasi, 'Dilasi'),
        (hasil_opening, 'Opening'),
        (hasil_closing, 'Closing')
    ]

    plt.figure(figsize=(14, 9))

    for i, (citra, judul) in enumerate(daftar_hasil, start=1):
        plt.subplot(2, 3, i)
        plt.imshow(citra, cmap='gray')
        plt.title(judul)
        plt.axis('off')

    plt.tight_layout()
    plt.show()

else:
    print("Gambar ktp1.jpg tidak ditemukan.")
```

```python
# -------------------------------------------------------------------------
# 18. Dilasi Circular SE
# -------------------------------------------------------------------------
img_d = cv2.imread(img_ktp_path, cv2.IMREAD_GRAYSCALE)

if img_d is not None:
    _, thresh = cv2.threshold(
        img_d, 127, 225, cv2.THRESH_BINARY
    )

    plt.figure(figsize=(12, 5))

    plt.subplot(1, 3, 1)
    plt.imshow(img_d, cmap='gray')
    plt.title('Citra Awal')
    plt.axis('off')

    for i, w in enumerate([3, 5], start=2):
        SED = cv2.getStructuringElement(
            cv2.MORPH_ELLIPSE, (w, w)
        )
        imgD = cv2.dilate(thresh, SED, iterations=1)

        plt.subplot(1, 3, i)
        plt.imshow(imgD, cmap='gray')
        plt.title(f'Dilasi Circular {w}x{w}')
        plt.axis('off')

    plt.tight_layout()
    plt.show()
else:
    print("Gambar ktp1.jpg tidak ditemukan.")
```

```python
# -------------------------------------------------------------------------
# 23. Perbandingan Opening dan Closing
# -------------------------------------------------------------------------
img = cv2.imread(img_ktp_path, cv2.IMREAD_GRAYSCALE)

if img is not None:
    _, thresh = cv2.threshold(
        img, 127, 255, cv2.THRESH_BINARY
    )

    kernel3 = np.ones((3, 3), np.uint8)
    kernel5 = np.ones((5, 5), np.uint8)

    opening3 = cv2.morphologyEx(
        thresh, cv2.MORPH_OPEN, kernel3
    )
    opening5 = cv2.morphologyEx(
        thresh, cv2.MORPH_OPEN, kernel5
    )
    closing3 = cv2.morphologyEx(
        thresh, cv2.MORPH_CLOSE, kernel3
    )
    closing5 = cv2.morphologyEx(
        thresh, cv2.MORPH_CLOSE, kernel5
    )

    hasil = [
        (img, 'Citra Awal'),
        (opening3, 'Opening 3x3'),
        (opening5, 'Opening 5x5'),
        (closing3, 'Closing 3x3'),
        (closing5, 'Closing 5x5')
    ]

    plt.figure(figsize=(14, 8))

    for i, (citra, judul) in enumerate(hasil, start=1):
        plt.subplot(2, 3, i)
        plt.imshow(citra, cmap='gray')
        plt.title(judul)
        plt.axis('off')

    plt.tight_layout()
    plt.show()
else:
    print("Gambar ktp1.jpg tidak ditemukan.")
```

```python

# -------------------------------------------------------------------------
# 24. Tugas Praktikum: Meningkatkan Keterbacaan Citra KTP
# -------------------------------------------------------------------------

import cv2
import numpy as np
import matplotlib.pyplot as plt

img_ktp = cv2.imread(img_ktp_path, cv2.IMREAD_GRAYSCALE)

if img_ktp is None:
    print("Gambar ktp1.jpg tidak ditemukan. Periksa path file.")
else:
    # A. Thresholding Otsu
    _, binary_otsu = cv2.threshold(
        img_ktp,
        0,
        255,
        cv2.THRESH_BINARY + cv2.THRESH_OTSU
    )

    # B. Thresholding Adaptive
    binary_adaptive = cv2.adaptiveThreshold(
        img_ktp,
        255,
        cv2.ADAPTIVE_THRESH_GAUSSIAN_C,
        cv2.THRESH_BINARY,
        21,
        5
    )

    # C. Opening untuk mengurangi noise kecil
    kernel_open = cv2.getStructuringElement(
        cv2.MORPH_RECT, (3, 3)
    )
    hasil_opening = cv2.morphologyEx(
        binary_otsu,
        cv2.MORPH_OPEN,
        kernel_open
    )

    # D. Closing untuk menutup celah kecil
    kernel_close = cv2.getStructuringElement(
        cv2.MORPH_RECT, (3, 3)
    )
    hasil_closing = cv2.morphologyEx(
        binary_otsu,
        cv2.MORPH_CLOSE,
        kernel_close
    )

    # E. Top-hat pada citra grayscale
    kernel_tophat = cv2.getStructuringElement(
        cv2.MORPH_RECT, (15, 15)
    )
    hasil_tophat = cv2.morphologyEx(
        img_ktp,
        cv2.MORPH_TOPHAT,
        kernel_tophat
    )

    # F. Black-hat pada citra grayscale
    hasil_blackhat = cv2.morphologyEx(
        img_ktp,
        cv2.MORPH_BLACKHAT,
        kernel_tophat
    )

    # Menampilkan seluruh hasil
    daftar_hasil = [
        (img_ktp, "Citra KTP Asli"),
        (binary_otsu, "Threshold Otsu"),
        (binary_adaptive, "Adaptive Threshold"),
        (hasil_opening, "Opening 3x3"),
        (hasil_closing, "Closing 3x3"),
        (hasil_tophat, "Top-hat"),
        (hasil_blackhat, "Black-hat")
    ]

    plt.figure(figsize=(16, 12))

    for i, (citra, judul) in enumerate(daftar_hasil, start=1):
        plt.subplot(3, 3, i)
        plt.imshow(citra, cmap="gray")
        plt.title(judul)
        plt.axis("off")

    plt.tight_layout()
    plt.show()
```

Pada tugas praktikum Modul 9, dilakukan penerapan operasi morfologi pada citra KTP menggunakan Python, OpenCV, dan NumPy. Proses pengolahan meliputi thresholding, opening, closing, Top-hat, dan Black-hat.

Setiap metode menghasilkan perubahan yang berbeda pada citra. Opening digunakan untuk membantu mengurangi noise kecil, sedangkan closing digunakan untuk membantu menutup celah kecil pada objek. Top-hat dan Black-hat digunakan untuk menonjolkan perbedaan intensitas tertentu pada citra.

Hasil pengolahan dibandingkan dengan citra awal untuk menentukan metode yang paling sesuai dalam membantu keterbacaan KTP. Pemilihan metode akhir dilakukan berdasarkan hasil visual, dengan memperhatikan kejelasan tulisan dan terjaganya detail karakter.
