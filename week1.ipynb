```python
import numpy as np
import pandas as pd
import cv2 as cv
from google.colab.patches import cv2_imshow #for image display
from skimage import io
from skimage import transform
from PIL import Image
import matplotlib.pylab as plt
```

```python
urls = ["https://i.pinimg.com/736x/1f/c5/21/1fc521a86c9486dfb3e1d7b78cb96c98.jpg",
        "https://i.pinimg.com/736x/8e/34/77/8e34776197235c3e68bda255979b0f17.jpg",
        "https://i.pinimg.com/736x/5c/30/21/5c3021c7c0d6a9df2de99b70e3008ff0.jpg"]
for url in urls:
      image = io.imread(url)                          #read image
      image = cv.resize(image, (0,0), fx=0.5, fy=0.5) #rresize image to half size
      image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB) #convert color to rgb
      final_frame = cv.hconcat((image, image_2)) # concatenate image
      cv2_imshow(final_frame)  #show image
      print('\n')
```

```python
tinggi = image_2.shape[0]
lebar = image_2.shape[1]
print("resolusi image: tinggi x lebar = ",tinggi," x ", lebar)
cv2_imshow(image_2)

```

```python
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)


for y in range (lebar):
   image_3[int((tinggi)/2),y] = [255,255,255]

final_frame = cv.hconcat((image_2, image_3))
cv2_imshow(final_frame)
```

```python
tinggi, lebar = image.shape[:2]

image_2 = image.copy()
image_3 = image.copy()

for x in range(tinggi):
    image_2[x, int(lebar / 2)] = [255, 255, 255]

for x in range(min(tinggi, lebar)):
    image_3[x, x] = [255, 255, 255]

image_2 = cv.cvtColor(image_2, cv.COLOR_BGR2RGB)
image_3 = cv.cvtColor(image_3, cv.COLOR_BGR2RGB)

final_frame = cv.hconcat((image_2, image_3))

cv2_imshow(final_frame)
```

```python

image_2 = image.copy()
# Menentukan posisi garis
y = int(tinggi / 2)
x_awal = int(lebar / 2) - 50
x_akhir = int(lebar / 2) + 50

#garis horizontal
for x in range(x_awal, x_akhir):
    image_2[y, x] = [255, 255, 255]

#  BGR ke RGB
image_2 = cv.cvtColor(image_2, cv.COLOR_BGR2RGB)

# hasil
cv2_imshow(image_2)
```

```python

image_3 = image.copy()

# Menentukan koordinat kotak
x1 = int(lebar / 2)
y1 = int(tinggi / 2) - 50

x2 = x1 + 70
y2 = y1 + 70

# Membuat kotak putih
for y in range(y1, y2):
    for x in range(x1, x2):
        image_3[y, x] = [255, 255, 255]

# BGR ke RGB
image_3 = cv.cvtColor(image_3, cv.COLOR_BGR2RGB)

# Menampilkan hasil
cv2_imshow(image_3)
```
